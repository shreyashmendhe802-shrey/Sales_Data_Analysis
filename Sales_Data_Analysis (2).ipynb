```python
import pandas as pd
import matplotlib.pyplot as plt

print("Sales Data Analysis Project")
```

```python
data = {
    "Order_ID": [1001, 1002, 1003, 1004, 1005,
                 1006, 1007, 1008, 1009, 1010,
                 1011, 1012, 1013, 1014, 1015],

    "Product": ["Laptop", "Mouse", "Keyboard", "Monitor", "Laptop",
                "Headphones", "Mouse", "Monitor", "Keyboard", "Laptop",
                "Headphones", "Mouse", "Monitor", "Laptop", "Keyboard"],

    "Category": ["Electronics", "Accessories", "Accessories", "Electronics", "Electronics",
                 "Accessories", "Accessories", "Electronics", "Accessories", "Electronics",
                 "Accessories", "Accessories", "Electronics", "Electronics", "Accessories"],

    "City": ["Nagpur", "Pune", "Mumbai", "Nagpur", "Pune",
             "Mumbai", "Nagpur", "Pune", "Mumbai", "Nagpur",
             "Pune", "Mumbai", "Nagpur", "Pune", "Mumbai"],

    "Quantity": [2, 5, 3, 2, 1, 4, 6, 2, 5, 1, 3, 7, 2, 2, 4],

    "Price": [55000, 800, 1500, 18000, 55000,
              2500, 800, 18000, 1500, 55000,
              2500, 800, 18000, 55000, 1500]
}

df = pd.DataFrame(data)

df
```

```python
df["Sales"] = df["Quantity"] * df["Price"]

total_sales = df["Sales"].sum()

print("Total Sales =", total_sales)
```

```python
product_sales = df.groupby("Product")["Sales"].sum().sort_values(ascending=False)

print(product_sales)
```

```python
product_sales.plot(kind="bar", figsize=(8,5))

plt.title("Sales by Product")
plt.xlabel("Product")
plt.ylabel("Sales (₹)")
plt.xticks(rotation=45)
plt.show()
```

```python
city_sales = df.groupby("City")["Sales"].sum().sort_values(ascending=False)

print(city_sales)
```

```python
city_sales.plot(kind="bar", figsize=(7,5))

plt.title("Sales by City")
plt.xlabel("City")
plt.ylabel("Sales (₹)")
plt.xticks(rotation=0)
plt.show()
```

```python
print("FINAL INSIGHTS")
print("-------------------------")

print("1. Total Sales: ₹", total_sales)

print("2. Highest Selling Product:",
      product_sales.idxmax(),
      "with sales of ₹", product_sales.max())

print("3. Lowest Selling Product:",
      product_sales.idxmin(),
      "with sales of ₹", product_sales.min())

print("4. Highest Sales City:",
      city_sales.idxmax(),
      "with sales of ₹", city_sales.max())

print("5. Lowest Sales City:",
      city_sales.idxmin(),
      "with sales of ₹", city_sales.min())
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Sales"], bins=5)

plt.title("Sales Distribution")
plt.xlabel("Sales (₹)")
plt.ylabel("Number of Orders")

plt.show()
```

```python
category_sales = df.groupby("Category")["Sales"].sum()
print(category_sales)

category_sales.plot(kind="bar", title="Category-wise Sales")
plt.ylabel("Sales")
plt.show()
```
